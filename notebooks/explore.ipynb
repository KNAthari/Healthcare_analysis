```python
import pandas as pd
import numpy as np 
import matplotlib as plt
from pathlib import Path

BASE_DIR = Path.cwd()
DATA_PATH = BASE_DIR.parent / "Data" / "healthcare_dataset_processed.csv"

df = pd.read_csv(DATA_PATH)

df.info()
df.head()
```

```python
# Summary of numerical data
df.describe()
```

```python
# Distribution of categorical data 
for col in ['Gender', 'Admission Type', 'Medical Condition', 'Insurance Provider', 'Blood Type']:
    print(f"\n--- {col} ---")
    print(df[col].value_counts())
```

```python
# Check for any null values
df.isnull().sum()
```

```python
# Length of Stay
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'], format="%Y-%m-%d")

df['Date of Admission'] = pd.to_datetime(df['Date of Admission'], format="%Y-%m-%d")

df["Length of Stay"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days

df["Length of Stay"].describe()
```
